# Create a Project




## Create a project file


```python
#@markdown ### Specify a filename for a new Project file, starting with `project`
project_filename = "project.csv" #@param {type:"string"}

project_card_fields = [
    'project_id',
    'state',
    'project_type',
    'description',
    'date_open_planned',
    'date_horizon',
    'date_open_actual',
    'postconstruction_functional_class',
    'change_in_functional_class',
    'area_type',
    'facility_name',
    'segment_length_miles',
]

with open(project_filename, 'w') as myfile:
    wr = csv.writer(myfile)
    wr.writerow(project_card_fields)
    
#for f in project_card_fields: print(f)
```

## Create a project ID 

Project ID should be a unique value from any other cardsets that you might want to combine it with.  This checks to see if your proposed project id conflicts with a standard set of repositories, but feel free to add your own checks as well.

```python
project_id   = "abc123" #@param {type:"string"}

check_cardset_1 = "../examples" #@param {type:"string"}
check_cardset_2 = "" #@param {type:"string"}
check_cardset_3 = "" #@param {type:"string"}
check_cardset_4  = "" #@param {type:"string"}


cardsets_to_check = [check_cardset_1,
                     check_cardset_2,
                     check_cardset_3,
                     check_cardset_4]


cs_with_project_id = []
for cs_loc in cardsets_to_check:
  if cs_loc: 
    
    cs_loc_orig = cs_loc
    
    if cs_loc.find('github')>=0: 
      cs_loc = forecastcards.github_url_to_dict(cs_loc)
    
    cs = forecastcards.Cardset(data_loc=cs_loc,validate=False)
    
    if project_id in cs.projects: 
      cs_with_project_id.append(cs_loc_orig)

if cs_with_project_id:
  print("!!! Select a different project ID. Duplicate Project ID found in following cardsets:","\n".join(cs_with_project_id))
else:
  print("Project ID good to go – not used in specified cardsets")
```

## Add Project variables

```python
state                       = ""      #@param {type:"string"}   # e.g. "KY"
description                 = ""      #@param {type:"string"}
project_type                = ""      #@param {type:"string"}
date_open_planned           = "YYYY-MM-DD"  #@param {type:"string"}
date_horizon                = "YYYY-MM-DD"  #@param {type:"string"}
date_open_actual            = "YYYY-MM-DD"  #@param {type:"string"}
postconstruction_functional_class = ""  #@param ["freeway", "principal arterial", "minor arterial", "major collector", "minor collector", "local", "ramp"] {allow-input: true}
change_in_functional_class  = 0       #@param {type:"slider", min:0, max:1, step:1}  # 1=Yes, 0=No
area_type                   = "urban" #@param ["urban", "rural"]
facility_name               = ""      #@param {type:"string"}
segment_length_miles        = 0.0     #@param {type:"number"}    # miles

if date_open_planned   == "YYYY-MM-DD": date_open_planned   = None
if date_horizon        == "YYYY-MM-DD": date_horizon        = None
if date_open_actual    == "YYYY-MM-DD": date_open_actual    = None

project_var_row = [
    project_id,
    state,
    project_type,
    description,
    date_open_planned,
    date_horizon,
    date_open_actual,
    postconstruction_functional_class,
    change_in_functional_class,
    area_type,
    facility_name,
    segment_length_miles,
]

for var_name, var_value in zip(project_card_fields, project_var_row):
    print(var_name, ": ", var_value)
```

## Review Project values before writing to CSV

## Write Project CSV

```python
with open(project_filename, 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(project_card_fields)
    writer.writerow([str(v) if v is not None else 'None' for v in project_var_row])
```
